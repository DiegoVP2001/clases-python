```python
#@title 🔧 Ejecuta esta celda primero (deja listas las funciones de hoy; no hace falta editarla)
import random
from collections import Counter

# Nuestra «base de datos» de juguete: el chat del curso
texto_base = "nos vemos en el patio . nos vemos en el patio . nos vemos en la plaza . ya llegué al casino ."


def _contar_siguientes(texto):
    palabras = texto.split()
    siguientes = {}
    for palabra, siguiente in zip(palabras, palabras[1:]):
        siguientes.setdefault(palabra, Counter())[siguiente] += 1
    return siguientes


def _conocidas(texto):
    return sorted(set(texto.split()))


def _sacar_ficha(tabla, conocidas, palabra):
    # Con bolsa propia: se saca una ficha al azar (más fichas = más probable).
    # Palabra que nunca vio: se saca al azar entre las palabras que sí conoce.
    if palabra in tabla:
        opciones = tabla[palabra]
        return random.choices(list(opciones), weights=list(opciones.values()))[0]
    return random.choice(conocidas)


def ver_tokens(texto):
    tokens = texto.split()
    print("Texto:", texto)
    print("Tokens (en nuestro mini modelo, cada palabra es un token):")
    print(" | ".join(tokens))
    print("Cantidad de tokens:", len(tokens))


def ver_tokens_modelos_reales():
    print("Ilustración: así podría cortar un modelo grande las palabras largas o poco comunes")
    ejemplos = [("casa", ["casa"]), ("inteligencia", ["inteli", "gencia"]), ("murciélago", ["mur", "ciél", "ago"])]
    for palabra, trozos in ejemplos:
        plural = "s" if len(trozos) > 1 else ""
        print(f"  {palabra:14s} → {' | '.join(trozos)}   ({len(trozos)} token{plural})")
    print("(Cada modelo corta a su manera: esto es solo una ilustración.)")


def mostrar_probabilidades(texto, palabra):
    tabla = _contar_siguientes(texto)
    if palabra not in tabla:
        print(f"«{palabra}» nunca apareció en el texto: no tiene bolsa propia.")
        print(f"Si le preguntamos qué viene después, el modelo saca una ficha AL AZAR entre las {len(_conocidas(texto))} palabras que sí conoce.")
        return
    total = sum(tabla[palabra].values())
    print(f"Bolsa de «{palabra}»: {total} fichas")
    for siguiente, veces in tabla[palabra].most_common():
        print(f"  «{siguiente}»: {veces} de {total} = {veces / total:.0%}")


def sacar_fichas(texto, palabra, cantidad=1):
    tabla = _contar_siguientes(texto)
    conocidas = _conocidas(texto)
    sacadas = [_sacar_ficha(tabla, conocidas, palabra) for _ in range(cantidad)]
    if palabra in tabla:
        print(f"Saqué {cantidad} ficha(s) de la bolsa de «{palabra}»:")
    else:
        print(f"«{palabra}» nunca apareció: saqué {cantidad} ficha(s) AL AZAR entre las {len(conocidas)} palabras que sí conoce:")
    print("  " + " ".join(sacadas))
    if cantidad > 1:
        conteo = Counter(sacadas)
        print("  Conteo: " + ", ".join(f"«{p}» {n}" for p, n in conteo.most_common()))


def _frase(texto, inicio, largo):
    tabla = _contar_siguientes(texto)
    conocidas = _conocidas(texto)
    frase = [inicio]
    while len(frase) < largo:
        frase.append(_sacar_ficha(tabla, conocidas, frase[-1]))
    return " ".join(frase)


def generar_frase(texto, inicio, largo=10):
    print(_frase(texto, inicio, largo))


def generar_varias(texto, inicio, largo=10, cantidad=5):
    for numero in range(1, cantidad + 1):
        print(f"{numero}: {_frase(texto, inicio, largo)}")


print("Funciones listas ✅")
```

### 💡 1. ¿Qué es un modelo de lenguaje (LLM)?

- **LLM** viene del inglés **L**arge **L**anguage **M**odel: *large* = grande · *language* = lenguaje · *model* = modelo.
- Es un modelo que aprendió, leyendo textos, a **predecir la siguiente palabra**.
- Es **estocástico**: su comportamiento **no se puede predecir con total exactitud**; o sea, tiene **azar** 🎲.
- El teclado del celular es un LLM **chiquito**; **ChatGPT** es uno **gigante**. **Los dos tienen azar**: lo que responden depende del azar y de las probabilidades con las que se entrenaron.
- Hoy usamos uno de juguete que aprende de un texto cortito: nuestra **«base de datos»**.

> **Idea clave:** un LLM (Large Language Model) aprende de textos a predecir la siguiente palabra y es estocástico: lo que responde tiene azar.

### 💡 3. Probabilidad y azar: la bolsa de fichas

> **Las respuestas de un LLM son azarosas.** El proceso para verlo es este: el modelo guarda, para cada palabra, una **bolsa de fichas** con lo que le siguió; con esas bolsas calcula **probabilidades**; y para responder **saca una ficha al azar**. Lo vemos paso a paso con nuestro mini modelo.

**① Aprender = contar**
- Por cada palabra, el modelo arma una **bolsa**.
- Cada vez que una palabra siguió a otra, mete una **ficha**.
- La bolsa de `en` tiene **2 fichas `el`** y **1 ficha `la`**.

**② Probabilidad = proporción de fichas**
- `el`: 2 de 3 ≈ **67 %** · `la`: 1 de 3 ≈ **33 %**.
- Si cambia el texto con el que se entrenó, cambian las proporciones.

**③ Azar = sacar una ficha sin mirar**
- El modelo **saca una ficha al azar** de la bolsa: `el` tiene más chances, pero **cualquiera puede salir** (`la` también).
- Por eso es **estocástico**: lo que responde es **azaroso** y depende de las **probabilidades con las que se entrenó**.

**④ Una palabra que nunca vio**
- No tiene bolsa propia (por ejemplo `biblioteca`).
- No se queda callado: **saca una ficha al azar entre las palabras que sí conoce**.
- Responde cualquier cosa de lo que sí sabe decir, sin ninguna base.

> **Idea clave:** la probabilidad es la proporción de fichas de cada palabra en la bolsa; el modelo saca una ficha al azar, así que lo que responde depende del azar y de las probabilidades con que se entrenó (estocástico). Si nunca vio la palabra, saca al azar entre las que sí conoce.

### 📌 Entonces, ¿qué debe quedar claro en tu cuaderno?

| Concepto | Qué significa |
|---|---|
| LLM (*Large Language Model*) | modelo de lenguaje que aprendió de textos a predecir la siguiente palabra y que es **estocástico** (tiene azar) |
| Token | un trozo de texto (una ficha) que el modelo lee como una sola pieza |
| Bolsa y probabilidad | por cada palabra, una bolsa con fichas de lo que le siguió; probabilidad = proporción de fichas (2 de 3 ≈ 67 %) |
| Estocástico | sistema cuyo comportamiento **no se puede predecir con total exactitud**: el modelo saca una ficha **al azar**, así que lo que responde depende del azar y de las probabilidades con que se entrenó |
| Palabra nunca vista | no tiene bolsa propia: saca al azar entre las palabras que sí conoce |
| Evidencia en la práctica | la misma pregunta a ChatGPT dos veces da respuestas distintas |
| Determinístico | lo contrario: misma entrada, **siempre** la misma salida (como una calculadora) |
