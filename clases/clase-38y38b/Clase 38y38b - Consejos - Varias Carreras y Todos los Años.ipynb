```python
import pandas as pd
import matplotlib.pyplot as plt

BASE = "https://raw.githubusercontent.com/DiegoVP2001/clases-python/master/clases/clase-38y38b/datos-compartidos-v3/"

sies = pd.read_csv(BASE + "01_sies.csv", low_memory=False)       # carreras e instituciones
conaset = pd.read_csv(BASE + "07_conaset_rm.csv")                # siniestros de tránsito RM (2020-2025)
print("SIES:", sies.shape, "| CONASET:", conaset.shape)
```

## Caso 2 · Varias carreras a la vez

Dos cambios:
1. `VALOR` pasa de texto a **lista** (entre corchetes, cada carrera entre comillas y separada por coma).
2. En la receta, `== VALOR` se cambia por `.isin(VALOR)`.

⚠️ Cada carrera debe estar escrita **igual que en la base**.

```python
COL_FILTRO = "Nombre carrera genérica"
VALOR      = ["Medicina", "Enfermería", "Kinesiología"]    # ← ahora es una lista
COL_GRUPO  = "Nombre de institución"
COL_NUMERO = "Ingreso_punto_medio"
RESUMEN    = "mean"
N          = 5
TITULO     = "Ingreso promedio por institución (Medicina, Enfermería y Kinesiología juntas)"

# ── La receta (no la cambies, salvo la línea del filtro que te indiquemos) ──
filtrada = sies[sies[COL_FILTRO].isin(VALOR)]      # ← .isin() en vez de ==

print("Filas que dejó el filtro:", len(filtrada))
if len(filtrada) == 0:
    print("⚠️ El filtro dejó 0 filas: revisa cómo escribiste VALOR (mayúsculas, tildes, comillas).")
else:
    resumen = filtrada.groupby(COL_GRUPO)[COL_NUMERO].agg(RESUMEN)
    top = resumen.sort_values(ascending=False).head(N)
    print(top.round(2))
    top.sort_values().plot(kind="barh", title=TITULO, xlabel=COL_NUMERO, ylabel=COL_GRUPO)
    plt.show()
```

### Caso 2b · Comparar las carreras *entre sí*

En el Caso 2 las barras son instituciones y mezclan las tres carreras. Si lo que quieres es comparar **las carreras**, cambia `COL_GRUPO`:

```python
COL_FILTRO = "Nombre carrera genérica"
VALOR      = ["Medicina", "Enfermería", "Kinesiología"]
COL_GRUPO  = "Nombre carrera genérica"                     # ← ahora las barras son carreras
COL_NUMERO = "Ingreso_punto_medio"
RESUMEN    = "mean"
N          = 5
TITULO     = "Ingreso promedio: Medicina vs Enfermería vs Kinesiología"

# ── La receta (no la cambies, salvo la línea del filtro que te indiquemos) ──
filtrada = sies[sies[COL_FILTRO].isin(VALOR)]

print("Filas que dejó el filtro:", len(filtrada))
if len(filtrada) == 0:
    print("⚠️ El filtro dejó 0 filas: revisa cómo escribiste VALOR (mayúsculas, tildes, comillas).")
else:
    resumen = filtrada.groupby(COL_GRUPO)[COL_NUMERO].agg(RESUMEN)
    top = resumen.sort_values(ascending=False).head(N)
    print(top.round(2))
    top.sort_values().plot(kind="barh", title=TITULO, xlabel=COL_NUMERO, ylabel=COL_GRUPO)
    plt.show()
```

## Caso 3 · Sumar todos los años (sin agrupar por año)

CONASET trae siniestros de 2020 a 2025. Si **no** quieres filtrar por año, `VALOR = None`: la receta usa la tabla completa y suma todos los años juntos.

```python
COL_FILTRO = "AÑO"            # no se usa, porque VALOR es None
VALOR      = None             # sin filtro: todos los años juntos
COL_GRUPO  = "COMUNA"
COL_NUMERO = "FALLECIDOS"
RESUMEN    = "sum"
N          = 5
TITULO     = "Fallecidos por comuna, suma 2020-2025 (RM)"

# ── La receta (no la cambies, salvo la línea del filtro que te indiquemos) ──
if VALOR == None:
    filtrada = conaset
else:
    filtrada = conaset[conaset[COL_FILTRO] == VALOR]

print("Filas que dejó el filtro:", len(filtrada))
if len(filtrada) == 0:
    print("⚠️ El filtro dejó 0 filas: revisa cómo escribiste VALOR (mayúsculas, tildes, comillas).")
else:
    resumen = filtrada.groupby(COL_GRUPO)[COL_NUMERO].agg(RESUMEN)
    top = resumen.sort_values(ascending=False).head(N)
    print(top.round(2))
    top.sort_values().plot(kind="barh", title=TITULO, xlabel=COL_NUMERO, ylabel=COL_GRUPO)
    plt.show()
```
