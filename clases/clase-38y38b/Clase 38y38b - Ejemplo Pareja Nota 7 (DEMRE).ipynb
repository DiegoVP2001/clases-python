### Paso 1 · Abro mi base y miro mis columnas

````python
import pandas as pd
import matplotlib.pyplot as plt

LINK = "https://raw.githubusercontent.com/DiegoVP2001/clases-python/master/clases/clase-38y38b/datos-compartidos-v3/02_demre.csv"

tabla = pd.read_csv(LINK)
print("Filas:", tabla.shape[0], "| Columnas:", tabla.shape[1])
tabla.head()
````

**Ficha de mis 5 columnas:**

````python
columnas = ["NOMBRE_CARRERA", "NOMBRE_UNIVERSIDAD", "PROM_OBLIGATORIAS_ULTIMO_SELEC", "VACANTES_REG", "COMPLETA_VACANTES_REG"]

for columna in columnas:
    print("Columna:", columna)
    print("  tipo:", tabla[columna].dtype)
    print("  valores distintos:", tabla[columna].nunique())
    print("  ejemplos:", list(tabla[columna].dropna().unique()[:3]))
    print("  vacíos:", tabla[columna].isna().sum())
````

### ⭐ Análisis 1 · ¿Qué universidades exigen el puntaje más alto para entrar a Medicina?

*(Paso 2: la receta con mis espacios llenos)*

````python
# ── Lo único que cambias (tu pregunta, traducida) ──
COL_FILTRO = "NOMBRE_CARRERA"                  # 🔎 ¿en qué columna miro?
VALOR      = "MEDICINA"                        # 🔎 ¿con qué valor me quedo?
COL_GRUPO  = "NOMBRE_UNIVERSIDAD"              # 🧺 ¿qué junto?
COL_NUMERO = "PROM_OBLIGATORIAS_ULTIMO_SELEC"  # 🔢 ¿qué cifra resumo?
RESUMEN    = "mean"                            # promedio
N          = 5                                 # ¿cuántos muestro?
TITULO     = "Puntaje del último seleccionado en Medicina (Admisión 2026)"

# ── La receta (no la cambies) ──
if VALOR == None:
    filtrada = tabla
else:
    filtrada = tabla[tabla[COL_FILTRO] == VALOR]

usadas = filtrada[COL_NUMERO].notna().sum()
perdidas = filtrada[COL_NUMERO].isna().sum()
print("Filas que dejó el filtro:", len(filtrada))
print("Con dato en", COL_NUMERO, ":", usadas, "| sin dato (se pierden):", perdidas)

if len(filtrada) == 0:
    print("⚠️ El filtro dejó 0 filas: revisa cómo escribiste VALOR (mayúsculas, tildes, comillas).")
else:
    resumen = filtrada.groupby(COL_GRUPO)[COL_NUMERO].agg(RESUMEN)
    top = resumen.sort_values(ascending=False).head(N)

    print("Cifras del top:")
    print(top.round(2))
    if RESUMEN != "mean":
        print("% del total:")
        print((top / resumen.sum() * 100).round(1))
    print("Mayor:", top.index[0], "→", round(top.iloc[0], 2))
    print("Menor del top:", top.index[-1], "→", round(top.iloc[-1], 2))
    print("Promedio de los", len(resumen), "grupos:", round(resumen.mean(), 2))

    if TITULO == "":
        TITULO = RESUMEN + " de " + COL_NUMERO + " por " + COL_GRUPO
    top.sort_values().plot(kind="barh", title=TITULO, xlabel=COL_NUMERO, ylabel=COL_GRUPO)
    plt.show()
````

**Paso 3 · Reviso lo que salió**

````python
# Paso 3 — Revisa tu resultado (solo ejecuta)
print("¿Cuántas filas hay detrás de cada barra?")
print(filtrada.groupby(COL_GRUPO)[COL_NUMERO].count().loc[top.index])
print()
print("Resumen de la columna que sumaste o promediaste:")
print(filtrada[COL_NUMERO].describe().round(2))
````

## 3️⃣ Mis análisis · 📽️ diapositivas 3, 4 y 5

### ⭐ Análisis 1 (principal)

**Mis respuestas (Análisis 1 ⭐):**

1. El ejemplo de DEMRE pregunta cuántas vacantes hay por área en carreras que no llenaron sus cupos. Nuestra pregunta es distinta en tres cosas: filtramos por **una carrera concreta** (`NOMBRE_CARRERA`) y no por si completó cupos, agrupamos por **universidad** en vez de por área, y resumimos con un **promedio** en vez de una suma. No pregunta cuánto hay, sino **qué tan exigente es** cada universidad.
2. 📽️ **Cómo lo investigamos.** (1) Nos quedamos solo con las filas de Medicina, porque mezclar todas las carreras habría comparado puntajes de cosas que no son comparables; quedaron 39 filas. (2) Juntamos esas filas por universidad, porque una misma universidad puede tener más de una fila de Medicina y queríamos **un solo número por universidad**. (3) De cada grupo calculamos el **promedio** del puntaje del último seleccionado, y no la suma, porque sumar puntajes no significa nada: preguntamos por lo típico de cada universidad, no por un total. (4) Ordenamos de mayor a menor y mostramos 5, porque con 29 universidades un gráfico sería ilegible.
3. 📽️ **Hallazgo y evidencia.** **La U. de Chile es la que más puntaje exige para entrar a Medicina, y los cinco primeros están muy cerca.** Su último seleccionado tuvo **982 puntos**; le siguen la Católica (958), la U. de Los Andes (948), Finis Terrae (932) y la U. de Valparaíso (927,25). Nos sorprendió que la U. de Los Andes y Finis Terrae, privadas, estén por sobre la U. de Santiago y la U. de Concepción: esperábamos que lo más alto fuera solo "las tradicionales".
4. 📽️ **Límites.** **No se perdió ninguna fila** (39 de 39 tenían dato), así que el filtro no descartó información. Pero detrás de cada barra hay **muy pocas filas**: las cuatro primeras universidades tienen **una sola fila** y la de Valparaíso tiene dos. Eso significa que cada barra es prácticamente el dato de **una** carrera, no el promedio de una universidad. Además, el puntaje es de **un solo año** (Admisión 2026) y mide al **último** seleccionado, no al promedio de los que entraron. NO podemos afirmar que "estudiar en la U. de Chile es mejor": solo que ahí **se necesitó más puntaje** ese año. Quien ignorara el límite podría pensar que la diferencia de 982 contra 958 es grande o permanente, cuando con una fila por universidad puede cambiar de un año a otro.
