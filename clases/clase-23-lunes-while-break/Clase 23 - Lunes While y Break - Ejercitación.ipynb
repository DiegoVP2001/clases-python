<summary>✅ Solución — Ejercicio guiado: Kiosco de sopaipillas del Aniversario</summary>

```python
sopaipillas = 12
while sopaipillas > 0:
    cantidad = int(input("¿Cuántas sopaipillas lleva? "))
    if cantidad == 0:
        print("Sin pedido, sigue la fila.")
        continue
    if cantidad > sopaipillas:
        respuesta = input("No alcanzan las que pide. ¿Lleva las que quedan? ")
        if respuesta == "no":
            print("Se cierra el kiosco con", sopaipillas, "sopaipillas.")
            break
        cantidad = sopaipillas
    sopaipillas = sopaipillas - cantidad
    print("Quedan", sopaipillas, "sopaipillas.")
```
