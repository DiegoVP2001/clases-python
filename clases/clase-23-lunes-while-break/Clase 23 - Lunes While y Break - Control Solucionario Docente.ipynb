## Ítem 1 — Batería del robot repartidor (25 pts)

**Concepto:** while

**Qué evalúa:** Repetir una cantidad de veces que no se conoce de antemano, con una variable de control que se actualiza en cada vuelta.

El robot que reparte libros en la biblioteca parte el día con 100 de batería. Cada entrega le descuenta una cantidad distinta, que el encargado va anotando, y el robot deja de aceptar entregas cuando la batería queda bajo 20.

**El programa debe:**
- Partir con la batería en 100.
- Ir pidiendo cuánto consumió cada entrega y descontarlo.
- Dejar de pedir entregas cuando la batería quede bajo 20.
- Mostrar con cuánta batería terminó.

```python
bateria = 100
while bateria >= 20:
    consumo = int(input("¿Cuánto consumió esta entrega? "))
    bateria = bateria - consumo
print("El robot se detuvo con", bateria, "de batería.")
```

| Una sola entrega grande | `85` | `El robot se detuvo con 15 de batería.` | 🔒 oculto |

| Salida final con el valor real | 5 | Imprime la batería que quedó, después del ciclo | 3 — imprime dentro del ciclo (repite el mensaje) | No imprime, o imprime un valor fijo |

## Ítem 2 — El primer video que despegó (30 pts)

**Concepto:** for + break

**Qué evalúa:** Cortar un recorrido apenas se encuentra lo que se buscaba, en vez de revisarlo todo.

Una cuenta de TikTok revisa las vistas de sus últimos 8 videos, uno por uno. Apenas aparece el primero que supera las 5.000 vistas, deja de revisar y avisa cuál fue.

**El programa debe:**
- Revisar hasta 8 videos, pidiendo las vistas de cada uno.
- Detenerse apenas uno supere las 5.000 vistas.
- Mostrar el número del video que lo logró.

```python
for numero_video in range(1, 9):
    vistas = int(input("¿Cuántas vistas tuvo este video? "))
    if vistas > 5000:
        print("El video", numero_video, "superó las 5.000 vistas.")
        break
```

### 🧪 Casos de prueba — Ítem 2

| Caso | Entradas | Salida esperada | |
|---|---|---|---|
| El tercero despega | `1200 / 3400 / 6100` | `El video 3 superó las 5.000 vistas.` | 👁️ visible |
| El primero despega | `7000` | `El video 1 superó las 5.000 vistas.` | 🔒 oculto |
| 5000 exactos no cuentan | `900 / 5000 / 800 / 5200` | `El video 4 superó las 5.000 vistas.` | 🔒 oculto |

### 🔍 Rúbrica parcelada — Ítem 2 (30 pts)

| Componente | Pts | ✅ Completo | ⚠️ Parcial | ❌ Cero |
|---|---|---|---|---|
| Recorrido de 8 videos | 6 | `for` con `range` que da 8 vueltas | 4 — cantidad distinta de vueltas, resto bien | Sin ciclo, o pide un solo dato |
| Lectura y conversión de cada dato | 4 | `input()` dentro del ciclo, convertido a número | 2 — sin convertir | Fuera del ciclo o valor fijo |
| Condición de detección | 8 | Compara contra 5000 con `>` | 6 — usa `>=` (el caso oculto de 5000 exactos lo detecta) · 4 — otro umbral, lógica bien planteada | Condición que no depende de las vistas |
| Corte efectivo con `break` | 8 | Corta apenas encuentra el primero | 4 — corta pero después de imprimir de más, o usa una bandera que igual detiene el recorrido | Sigue revisando los 8 (reporta el último, no el primero) |
| Mensaje identifica cuál fue | 4 | Imprime el número del video correcto | 2 — imprime las vistas en vez del número | No distingue cuál fue |

> **Nota de corrección:** El caso "ninguno de los 8 supera" no se exige — `for...else` está fuera del alcance del curso. No descuenta si el programa no lo cubre.

## Ítem 3 — Máquina de reciclaje (45 pts)

**Concepto:** while + continue

**Qué evalúa:** Combinar un ciclo que se repite hasta cumplir una meta con el salto de los casos que no aportan.

La máquina de reciclaje del supermercado suma 5 puntos por cada botella y entrega el vale al llegar a 20 puntos. Si le meten algo que no es botella, lo devuelve, no suma nada y queda esperando lo siguiente.

**El programa debe:**
- Ir pidiendo qué se ingresó a la máquina (`botella` u otra cosa).
- Sumar 5 puntos solo por las botellas.
- Dejar de pedir al llegar a los 20 puntos y mostrar el total.

```python
puntos = 0
while puntos < 20:
    objeto = input("¿Qué ingresaste a la máquina? ")
    if objeto != "botella":
        continue
    puntos = puntos + 5
print("Vale entregado con", puntos, "puntos.")
```

### 🧪 Casos de prueba — Ítem 3

| Caso | Entradas | Salida esperada | |
|---|---|---|---|
| Con una lata entremedio | `botella / lata / botella / botella / botella` | `Vale entregado con 20 puntos.` | 👁️ visible |
| Solo botellas | `botella / botella / botella / botella` | `Vale entregado con 20 puntos.` | 🔒 oculto |
| Empieza con dos objetos inválidos | `caja / caja / botella / botella / botella / botella` | `Vale entregado con 20 puntos.` | 🔒 oculto |

### 🔍 Rúbrica parcelada — Ítem 3 (45 pts)

| Componente | Pts | ✅ Completo | ⚠️ Parcial | ❌ Cero |
|---|---|---|---|---|
| Estructura del ciclo hasta la meta | 10 | `while` cuya condición depende de los puntos acumulados | 7 — usa `<=` u otro umbral, lógica bien planteada · 5 — `while True` con un corte equivalente que sí funciona | Cantidad fija de vueltas (`for`), o condición que no depende de los puntos |
| Lectura dentro del ciclo | 5 | `input()` dentro del `while` | 3 — lo pide una sola vez antes del ciclo pero el resto funciona | No pide el dato |
| Condición del caso a descartar | 8 | Compara contra `"botella"` correctamente | 5 — compara contra un booleano o un texto distinto pero apunta bien | Condición que no depende de lo ingresado |
| Salto efectivo (no suma y sigue) | 12 | Lo que no es botella no suma puntos y el ciclo continúa. **Resolverlo con `if/else` en vez de `continue` vale el puntaje completo** — es un equivalente legítimo y aquí se califica el comportamiento, no la forma | 6 — a medias: no suma pero rompe el ciclo (`break`), o suma igual pero avisa | Suma puntos por cualquier cosa |
| Acumulación de los válidos | 6 | Suma 5 por botella | 3 — suma otro valor de forma consistente | No acumula |
