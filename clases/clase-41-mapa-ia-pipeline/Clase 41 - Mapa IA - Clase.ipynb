## 📘 Concepto 1 · El mapa de la IA

El mapa de la IA

### Regla y patrón

> **Idea clave — Regla:** instrucción que escribe una persona y que le dice a la máquina qué hacer en cada caso. No cambia sola.

*Ejemplos:* «si hace calor, enciende el ventilador»; una calculadora.

> **Idea clave — Patrón:** algo que se repite en los datos y que la máquina descubre sola, mirando muchos ejemplos.

*Ejemplos:* con calor se venden más helados; quienes ven un tipo de video suelen ver otros parecidos.

👉 **Vamos a la actividad M1 de la página.**

### IA, ML, DL e IA generativa

> **Idea clave — IA:** máquinas que hacen tareas que normalmente requieren inteligencia humana: aprender, razonar y decidir.

*Ejemplos:* el GPS que calcula la ruta más corta; el rival de ajedrez del computador.

> **Idea clave — Machine Learning (ML):** la computadora descubre patrones en los datos y decide sin que le indiquen cómo.

*Ejemplos:* TikTok sugiere videos según lo que vemos; el banco detecta una compra rara con la tarjeta; predecir si hay que regar una viña mañana.

> **Idea clave — Deep Learning (DL):** ML con redes neuronales de muchas capas; sirve para imágenes, voz y texto.

*Ejemplos:* los subtítulos automáticos de YouTube; un traductor de voz; un auto que reconoce peatones en la cámara.

> **Idea clave — IA generativa:** produce contenido nuevo (texto, imágenes, música) a partir de lo que aprendió. Hoy se construye sobre todo con DL (por ejemplo, ChatGPT).

*Ejemplos:* ChatGPT escribe un cuento; una app crea una canción a partir de una descripción.

## 📘 Concepto 2 · El recorrido de los datos

Caso del diagrama: riego automatizado de una viña de Isla de Maipo, como Santa Ema (datos ficticios).

El recorrido de los datos

### Los 7 pasos

1. Definir el problema (¿hay que regar mañana?).
2. Reunir los datos y reconocer variables y etiqueta.
3. Separar los datos en dos sets: el set de entrenamiento y el set de prueba.
4. Entrenar el modelo con el set de entrenamiento.
5. Predecir sobre el set de prueba.
6. Evaluar cuántas veces acertó.
7. Usar el modelo en casos nuevos y revisar.

En las actividades de la página cambiamos de ejemplo: el sistema de climatización de una bodega de fermentación.

👉 **Vamos a la actividad P1 de la página** (cambiamos a la **página 2**: se abre sola al terminar la actividad anterior).

### Variables y etiqueta

> **Idea clave — Variables y etiqueta:** las variables son lo que el modelo recibe; la etiqueta es lo que queremos que prediga.

👉 **Vamos a la actividad P2 de la página.**

### Entrenar y predecir

> **Idea clave — Entrenar y predecir:** entrenar es aprender de ejemplos que traen su respuesta; predecir es usar lo aprendido en un caso.

> **Idea clave — Set de entrenamiento:** es donde la computadora aprende: ve las variables y la etiqueta.

> **Idea clave — Set de prueba:** es donde vemos si es capaz de predecir: no ve la etiqueta cuando responde.

👉 **Vamos a la actividad P3 de la página.**

### El set de prueba se guarda aparte

> **Idea clave — Prueba aparte:** si evaluamos con lo mismo que entrenamos, no sabemos cómo le irá con casos nuevos.
