## Parte A · Las etapas del pipeline

> **Idea clave — Pipeline (flujo de trabajo):** un modelo de ML se construye en etapas, siempre en el mismo orden: **data → features y label → train/test split → algorithm → model**.

Completa la tabla con tus palabras mientras se explica.

| Etapa | En español | ¿Qué hago en esta etapa? |
|---|---|---|
| Data | Datos | |
| Features / Label | Variables de entrada / etiqueta a predecir | |
| Train/test split | División entrenamiento / prueba | |
| Algorithm | Algoritmo | |
| Model | Modelo | |

> 🧑‍🏫 **PARA TI (no se muestra al curso)**
>
> **TU EXPLICACIÓN (12 min)**
> - Usa SIEMPRE el mismo ejemplo de punta a punta: *predecir si un estudiante viene caminando al liceo según los minutos de trayecto*.
> - Feature = lo que **conozco** (minutos de trayecto). Label = lo que quiero **predecir** (¿viene caminando?).
> - Train = lo que el modelo ve para aprender. Test = lo que se **guarda** para ver si realmente aprendió.
> - Analogía útil para el Ticket: estudiar con la prueba de ayer vs. rendir una prueba nueva.
> - Aclara que en la práctica hay más de una feature; aquí usamos una para poder hacerlo a mano.
