## 🎯 Objetivo

Consolidar el uso de `for` junto con `range()` y de ciclos `for` anidados, reconociendo cuándo un problema necesita un solo ciclo y cuándo necesita uno dentro de otro, resolviendo variaciones distintas a las vistas en clase.

---

## 📌 Resumen rápido

**`for` + `range()`**

Repite un bloque una vez por cada valor de `range(inicio, fin, salto)`. El `fin` nunca se incluye.

```python
for numero in range(1, 4):
    print(numero)
# >> 1
# >> 2
# >> 3
```

**`for` anidado**

Un `for` dentro de otro: por cada vuelta del externo, el interno corre completo antes de que el externo avance.

```python
for fila in range(1, 3):
    for columna in range(1, 3):
        print(fila, columna)
# >> 1 1
# >> 1 2
# >> 2 1
# >> 2 2
```
