## 🔁 Ejercicio guiado — Riego por goteo en una parcela

En una parcela de Isla de Maipo el sistema de riego está dividido en canales, y cada canal alimenta varios surcos de viña. El encargado quiere que el programa recorra canal por canal, revisando surco por surco, para confirmar que el agua llegó a todos antes de la temporada seca.

**El programa debe:**
- Recorrer los 3 canales de la parcela.
- Por cada canal, recorrer sus 4 surcos y mostrar cuál canal y cuál surco se está revisando.

**Resultado esperado:**
```
Canal 1 - Surco 1
Canal 1 - Surco 2
Canal 1 - Surco 3
Canal 1 - Surco 4
Canal 2 - Surco 1
...
Canal 3 - Surco 4
```

````python
# ✅ Solución esperada
canales = 3
surcos_por_canal = 4

for canal in range(1, canales + 1):
    for surco in range(1, surcos_por_canal + 1):
        print("Canal", canal, "- Surco", surco)
````

### Pregunta 1

```python
for numero in range(3, 10, 3):
    print(numero)
```

¿Qué valores imprime este código?

- A: 3, 6, 9
- B: 3, 6, 9, 10
- C: 3, 6
- D: 0, 3, 6, 9

### Pregunta 3


- B: 4 veces
- C: 6 veces
