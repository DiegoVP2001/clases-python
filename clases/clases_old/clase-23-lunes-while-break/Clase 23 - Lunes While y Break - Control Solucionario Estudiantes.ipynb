## Ítem 1 — Batería del robot repartidor (25 pts)

El robot que reparte libros en la biblioteca parte el día con 100 de batería. Cada entrega le descuenta una cantidad distinta, que el encargado va anotando, y el robot deja de aceptar entregas cuando la batería queda bajo 20.

**El programa debe:**
- Partir con la batería en 100.
- Ir pidiendo cuánto consumió cada entrega y descontarlo.
- Dejar de pedir entregas cuando la batería quede bajo 20.
- Mostrar con cuánta batería terminó.

```python
bateria = 100
while bateria >= 20:
    consumo = int(input("¿Cuánto consumió esta entrega? "))
    bateria = bateria - consumo
print("El robot se detuvo con", bateria, "de batería.")
```

🔎 **Qué se revisó:** Repetir una cantidad de veces que no se conoce de antemano, con una variable de control que se actualiza en cada vuelta.

## Ítem 2 — El primer video que despegó (30 pts)

Una cuenta de TikTok revisa las vistas de sus últimos 8 videos, uno por uno. Apenas aparece el primero que supera las 5.000 vistas, deja de revisar y avisa cuál fue.

**El programa debe:**
- Revisar hasta 8 videos, pidiendo las vistas de cada uno.
- Detenerse apenas uno supere las 5.000 vistas.
- Mostrar el número del video que lo logró.

```python
for numero_video in range(1, 9):
    vistas = int(input("¿Cuántas vistas tuvo este video? "))
    if vistas > 5000:
        print("El video", numero_video, "superó las 5.000 vistas.")
        break
```

🔎 **Qué se revisó:** Cortar un recorrido apenas se encuentra lo que se buscaba, en vez de revisarlo todo.

## Ítem 3 — Máquina de reciclaje (45 pts)

La máquina de reciclaje del supermercado suma 5 puntos por cada botella y entrega el vale al llegar a 20 puntos. Si le meten algo que no es botella, lo devuelve, no suma nada y queda esperando lo siguiente.

**El programa debe:**
- Ir pidiendo qué se ingresó a la máquina (`botella` u otra cosa).
- Sumar 5 puntos solo por las botellas.
- Dejar de pedir al llegar a los 20 puntos y mostrar el total.

```python
puntos = 0
while puntos < 20:
    objeto = input("¿Qué ingresaste a la máquina? ")
    if objeto != "botella":
        continue
    puntos = puntos + 5
print("Vale entregado con", puntos, "puntos.")
```

🔎 **Qué se revisó:** Combinar un ciclo que se repite hasta cumplir una meta con el salto de los casos que no aportan.
