## 🎯 Objetivo

Consolidar el patrón definir → llamar → guardar el resultado de una función en una variable → reutilizarlo con `return`, sin confundirlo con lo que muestra un `print()`, antes del control individual de la segunda parte de la clase.

## 💡 Propósito

> Confiar en el proceso es apoyarte en un método que ya practicaste, en vez de improvisar cuando el tiempo apura. Hoy lo practicas siguiendo siempre el mismo orden al usar una función — definir, consultar los datos, evaluar guardando el resultado en una variable, y recién ahí imprimir o reutilizar ese resultado — incluso en ejercicios cortos donde tentaría saltárselo.

No lleva nota: lo que se evalúa hoy es el Control de la segunda parte.

💡 **Dos cosas que conviene tener frescas antes de partir:**

- Solo `return` entrega el valor de vuelta a quien llama la función; un `print()` dentro de la función lo muestra en pantalla, pero no lo deja disponible para usarlo después — si tu función no tiene `return`, lo que guardes será `None`.
- El orden **definir → consultar los datos → evaluar la función guardando su resultado en una variable → imprimir** ordena cualquier programa que use una función. No te lo saltes aunque el ejercicio sea corto.

```python
def descuento_por_entradas(entradas_compradas):
    if entradas_compradas >= 8:
        return 4000
    elif entradas_compradas >= 4:
        return 1500
    return 0

entradas = int(input("¿Cuántas entradas compra el grupo? "))
precio_unitario = int(input("¿Cuál es el precio de cada entrada? "))
total_bruto = entradas * precio_unitario

descuento = descuento_por_entradas(entradas)
total_final = total_bruto - descuento
print("Total a pagar:", total_final)
```
