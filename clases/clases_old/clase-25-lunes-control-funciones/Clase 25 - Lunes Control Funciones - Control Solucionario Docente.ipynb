## Ítem 1 — Piscina municipal de Isla de Maipo en verano (25 pts)

**Concepto:** función con return, guardar en variable e imprimir

**Qué evalúa:** Definir una función simple con `return`, guardarla en una variable y usar ese valor guardado para imprimir — sin caer en llamar la función directo dentro del `print()` ni en usar `print()` en vez de `return` dentro de la función.

La piscina municipal de Isla de Maipo tiene un aforo máximo fijado por seguridad, y el guardavidas quiere saber al instante cuántos cupos quedan disponibles cada vez que entra o sale gente, sin tener que calcularlo a mano.

**El programa debe:**
- Definir una función que reciba el aforo máximo y la cantidad de personas que están dentro, y devuelva cuántos cupos quedan disponibles con `return`.
- Pedir esos dos datos.
- Llamar la función, guardar el resultado en una variable, y mostrar cuántos cupos quedan disponibles.

```python
def cupos_disponibles(aforo_maximo, personas_dentro):
    return aforo_maximo - personas_dentro

aforo = int(input("¿Cuál es el aforo máximo de la piscina? "))
personas = int(input("¿Cuántas personas hay dentro ahora? "))

cupos = cupos_disponibles(aforo, personas)
print("Cupos disponibles:", cupos)
```

## Ítem 2 — Bingo solidario del Centro de Padres (30 pts)

**Concepto:** guardar el resultado de una función y reutilizarlo en un cálculo posterior

**Qué evalúa:** Guardar el resultado de una función en una variable y reutilizarlo en un cálculo posterior (no solo imprimirlo directo) — el punto de quiebre central de la clase de uso del retorno.

El Centro de Padres organiza un bingo solidario: venden cartones a un precio fijo, y al cerrar la jornada quieren saber cuánto les quedó de ganancia después de pagar el arriendo del salón.

**El programa debe:**
- Definir una función que reciba la cantidad de cartones vendidos y el precio de cada cartón, y devuelva el ingreso total con `return`.
- Pedir esos dos datos, llamar la función y guardar el ingreso en una variable.
- Pedir cuánto costó el arriendo del salón y restárselo al ingreso guardado para obtener la ganancia.
- Mostrar la ganancia final.

```python
def ingreso_bingo(cartones_vendidos, precio_carton):
    return cartones_vendidos * precio_carton

cartones = int(input("¿Cuántos cartones se vendieron? "))
precio_carton = int(input("¿Cuál es el precio de cada cartón? "))
ingreso = ingreso_bingo(cartones, precio_carton)

arriendo = int(input("¿Cuánto costó el arriendo del salón? "))
ganancia = ingreso - arriendo
print("Ganancia del bingo:", ganancia)
```

## Ítem 3 — Pedido de arcilla para el taller de cerámica (45 pts)

**Concepto:** función con if/elif/else que retorna un descuento, combinado con un total calculado aparte

**Qué evalúa:** Combinar una función con retorno condicional (3 tramos) con un cálculo aparte (total bruto), reutilizando el valor guardado de la función en una resta posterior — el ítem de cierre, que junta el patrón de guardar/reutilizar de 24b con el retorno condicional de 24a.

El taller de cerámica compra arcilla por kilo para todo el semestre, y el proveedor ofrece un descuento total según la cantidad de kilos que pidan de una sola vez: mientras más kilos, mayor el descuento.

**El programa debe:**
- Definir una función que reciba la cantidad de kilos pedidos y devuelva el descuento que corresponde con `return`: \$5.000 de descuento si piden 20 kilos o más, \$2.000 si piden entre 10 y 19, y sin descuento si piden menos de 10.
- Pedir la cantidad de kilos y el precio de cada kilo, y calcular aparte el total bruto (kilos por precio).
- Llamar la función, guardar el descuento, y restárselo al total bruto para obtener el total final.
- Mostrar el total final a pagar.

```python
def descuento_por_kilos(kilos_pedidos):
    if kilos_pedidos >= 20:
        return 5000
    elif kilos_pedidos >= 10:
        return 2000
    return 0

kilos = int(input("¿Cuántos kilos de arcilla piden? "))
precio_kilo = int(input("¿Cuál es el precio por kilo? "))
total_bruto = kilos * precio_kilo

descuento = descuento_por_kilos(kilos)
total_final = total_bruto - descuento
print("Total a pagar:", total_final)
```
