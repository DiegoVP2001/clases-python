### Ejercicio 1 — Kilometraje semanal de natación (18 pts)

El club de natación de la piscina municipal registra cuántos kilómetros nada cada estudiante en la semana, para reconocer la constancia de cada nadador. Escribe el programa que, dado el kilometraje nadado, muestre el nivel correspondiente.

**El programa debe:**
- Pedir con `input()` los **kilómetros nadados esta semana** (**puede tener decimales**)
- Clasificar el kilometraje en estos 4 niveles:
  - Menos de 5 km → `"Nivel: Recién empezando."`
  - Entre 5 y menos de 15 km → `"Nivel: En camino."`
  - Entre 15 y menos de 25 km → `"Nivel: Buen ritmo."`
  - 25 km o más → `"Nivel: ¡Semana destacada!"`

</tr>
<tr>
  <td>📤 <em>El programa imprime</em><pre>Nivel: En camino.</pre></td>
  <td>📤 <em>El programa imprime</em><pre>Nivel: ¡Semana destacada!</pre></td>

```python
km_nadados_semana = float(input("Ingresa cuántos kilómetros nadaste esta semana: "))

if km_nadados_semana < 5:
    print("Nivel: Recién empezando.")
elif km_nadados_semana < 15:
    print("Nivel: En camino.")
elif km_nadados_semana < 25:
    print("Nivel: Buen ritmo.")
else:
    print("Nivel: ¡Semana destacada!")
```

🔎 **Qué se revisó:** Que las **4 categorías** quedaran bien delimitadas, sin huecos ni superposiciones en los límites entre un nivel y el siguiente.

Escribe el programa que, dado el porcentaje de batería, informe en qué modo queda el control.

### Ejercicio 3 — Entrada a la Fiesta de la Vendimia (24 pts)

La boletería de la Fiesta de la Vendimia de Isla de Maipo arma el acceso según el tipo de entrada. Para la entrada VIP, además revisa si la persona lleva un acompañante, porque eso cambia el acceso que recibe. Escribe el programa que, dado el tipo de entrada y —cuando corresponda— si lleva acompañante, muestre a qué acceso queda.

**El programa debe:**
- Pedir con `input()` el **tipo de entrada** (la persona responde **exactamente** una de estas palabras: "general", "preferencial" o "vip")
- Si el tipo es **"vip"**, pedir además con `input()` si **lleva acompañante** (responde **exactamente** "si" o "no") — esta pregunta no se hace para los otros tipos
- Mostrar el mensaje que corresponda a cada uno de estos casos:
  - tipo "general" → `"Acceso por la fila general."`
  - tipo "preferencial" → `"Acceso por la fila preferencial."`
  - tipo "vip" sin acompañante → `"Acceso VIP directo."`
  - tipo "vip" con acompañante → `"Acceso VIP directo con acompañante."`

```python
tipo_entrada = input("Ingresa tu tipo de entrada (general/preferencial/vip): ")

if tipo_entrada == "vip":
    lleva_acompanante = input("¿Llevas acompañante? (si/no): ")
    if lleva_acompanante == "si":
        print("Acceso VIP directo con acompañante.")
    else:
        print("Acceso VIP directo.")
elif tipo_entrada == "preferencial":
    print("Acceso por la fila preferencial.")
else:
    print("Acceso por la fila general.")
```

🔎 **Qué se revisó:** Que la pregunta por el acompañante solo se hiciera para el tipo "vip", y que las **cuatro** combinaciones posibles (general, preferencial, vip sin acompañante, vip con acompañante) mostraran su mensaje correcto.

### Ejercicio 4 — Convención de videojuegos y anime (16 pts)

Para entrar a una convención de videojuegos y anime, el sistema revisa primero si la persona ya compró su entrada digital; si no la compró, revisa si el efectivo que lleva alcanza para pagar la entrada en la puerta. Escribe el programa que, según esos datos, muestre cómo puede entrar la persona.

**El programa debe:**
- Pedir con `input()` si ya tiene **entrada digital comprada** (la persona responde **exactamente** "si" o "no")
- Si **no** tiene entrada digital, pedir además con `input()` **cuánto efectivo lleva** (un **número entero**, en pesos)
- Verificar primero la entrada digital; solo si no la tiene, verificar si el efectivo **alcanza** el costo de la entrada en la puerta (\$6.000)
- Mostrar el mensaje que corresponda a cada uno de estos tres caminos:
  - Tiene entrada digital → `"Ingresa directo mostrando su entrada digital."`
  - No tiene entrada Y el efectivo alcanza → `"Paga la entrada en la puerta."`
  - No tiene entrada Y el efectivo no alcanza → `"No le alcanza el efectivo para entrar."`

<table>
<tr>
  <th>Ejemplo 1</th>
  <th>Ejemplo 2</th>
</tr>
<tr>
  <td>📥 <em>El usuario ingresa</em><pre>si</pre></td>
  <td>📥 <em>El usuario ingresa</em><pre>no<br>4000</pre></td>
</tr>
<tr>
  <td>📤 <em>El programa imprime</em><pre>Ingresa directo mostrando su entrada digital.</pre></td>
  <td>📤 <em>El programa imprime</em><pre>No le alcanza el efectivo para entrar.</pre></td>
</tr>
</table>

```python
tiene_entrada_digital = input("¿Ya tienes tu entrada digital comprada? (si/no): ")

if tiene_entrada_digital == "si":
    print("Ingresa directo mostrando su entrada digital.")
else:
    efectivo_disponible = int(input("Ingresa cuánto efectivo llevas: "))
    if efectivo_disponible >= 6000:
        print("Paga la entrada en la puerta.")
    else:
        print("No le alcanza el efectivo para entrar.")
```

🔎 **Qué se revisó:** Que se verificara **primero** la entrada digital, que la pregunta por el efectivo solo apareciera cuando correspondía (sin entrada), y que los **tres** caminos posibles mostraran su mensaje exacto.

El **casillero inteligente** del gimnasio del liceo solo se abre cuando el **código ingresado** es exactamente igual al **código asignado** al estudiante.

**Ítem 2B.4** (4 pts)

Un **dron de reparto** revisa primero si tiene **batería suficiente** para volar, y solo entonces si **detectó la dirección de entrega**.

**Ítem 2B.5** (5 pts)

Una **tienda de artículos de skate** clasifica el nivel de **fidelidad** de un cliente según la cantidad de **compras acumuladas en el año**, de la categoría más exigente a la más básica.

```python
compras_acumuladas_anio = 12

if compras_acumuladas_anio >= 14:
    mensaje_fidelidad = "Cliente VIP."
elif compras_acumuladas_anio >= 7:
    mensaje_fidelidad = "Cliente frecuente."
elif compras_acumuladas_anio >= 3:
    mensaje_fidelidad = "Cliente ocasional."
else:
    mensaje_fidelidad = "Nuevo cliente."

print(mensaje_fidelidad)
# Esperado: Cliente frecuente.
```

🔎 **Qué se revisó:** Que las condiciones quedaran ordenadas de **mayor a menor** exigencia, para que ningún caso quedara mal clasificado por una condición más amplia evaluada antes de tiempo.
