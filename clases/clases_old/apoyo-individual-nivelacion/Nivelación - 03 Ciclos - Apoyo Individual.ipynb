## 🧩 Bloque 1 — `for` + `range()`

- `for` repite un bloque de código una vez por cada valor de una secuencia.
- `range(fin)` genera números empezando en 0 y terminando ANTES de `fin` (`fin` nunca se incluye).
- Todo lo que esté indentado bajo el `for` se repite en cada vuelta.
- Por defecto `range(fin)` siempre parte en 0 — en el Bloque 2 verás cómo elegir tú el punto de partida.

**Ejemplo resuelto:**

```python
for numero_vuelta in range(5):
    print("Vuelta número", numero_vuelta)
>> Vuelta número 0
>> Vuelta número 1
>> Vuelta número 2
>> Vuelta número 3
>> Vuelta número 4
```

## 🧩 Bloque 2 — `range(inicio, fin, salto)`

- `range(inicio, fin, salto)` permite elegir en qué número empezar, dónde detenerse (sin incluir `fin`) y de cuánto en cuánto avanzar.
- Si el salto es negativo, la cuenta va hacia atrás (cuenta regresiva).
- Si se omite `inicio`, parte en 0; si se omite `salto`, avanza de 1 en 1 (esto ya lo viste en el Bloque 1).

**Ejemplo resuelto:**

```python
for numero_partido in range(1, 8, 1):
    print("Partido número", numero_partido)
>> Partido número 1
>> Partido número 2
>> Partido número 3
>> Partido número 4
>> Partido número 5
>> Partido número 6
>> Partido número 7
```

## 🧩 Bloque 3 — Acumulador dentro de un ciclo

- Un acumulador es una variable que se inicializa ANTES del `for`, fuera del bloque indentado.
- Dentro del ciclo se le va sumando un valor en cada vuelta, así va guardando un total.
- ⚠️ Si el acumulador se inicializa DENTRO del ciclo, se reinicia en cada vuelta y nunca acumula de verdad.

**Ejemplo resuelto:**

```python
puntos_totales = 0
for numero_ronda in range(1, 5):
    puntos_totales = puntos_totales + 10
    print("Después de la ronda", numero_ronda, ":", puntos_totales, "puntos")
>> Después de la ronda 1 : 10 puntos
>> Después de la ronda 2 : 20 puntos
>> Después de la ronda 3 : 30 puntos
>> Después de la ronda 4 : 40 puntos
```

## 🧩 Bloque 4 — `for` anidado: filas y columnas

- Un `for` puede ir dentro de otro `for`: el ciclo interno corre completo por cada vuelta del ciclo externo.
- El ciclo externo suele representar las filas, y el interno las columnas.
- Cada nivel de anidamiento necesita su propia sangría: el cuerpo del `for` interno va indentado dentro del cuerpo del externo.

**Ejemplo resuelto:**

```python
for fila in range(1, 3):
    for asiento in range(1, 4):
        print("Fila", fila, "- Asiento", asiento)
>> Fila 1 - Asiento 1
>> Fila 1 - Asiento 2
>> Fila 1 - Asiento 3
>> Fila 2 - Asiento 1
>> Fila 2 - Asiento 2
>> Fila 2 - Asiento 3
```

## 🧩 Bloque 5 — Construir la salida por fila

- `print(..., end=" ")` mantiene los valores en la misma línea en vez de saltar de línea automáticamente.
- Un `print()` vacío después del ciclo interno (pero dentro del externo) salta a la siguiente línea al terminar cada fila.
- ⚠️ Sin ese `print()` vacío, todo quedaría impreso en una sola línea corrida.

**Ejemplo resuelto:**

```python
for fila in range(1, 3):
    for asiento in range(1, 4):
        print(asiento, end=" ")
    print()
>> 1 2 3
>> 1 2 3
```
