## 🧩 Bloque 1 — Variables y `print()`

- Una variable guarda un valor para reutilizarlo después. El nombre debe ser descriptivo y en español (`entradas_vendidas`, no `x`).
- `print()` muestra en pantalla texto y valores de variables — se pueden mezclar usando comas.
- El texto entre comillas (`"..."`) se muestra tal cual; una variable (sin comillas) muestra el valor que tiene guardado.

**Ejemplo resuelto:**

```python
nombre_feria = "Feria de Isla de Maipo"
puestos_instalados = 24
print("Bienvenido a la", nombre_feria)
print("Puestos instalados:", puestos_instalados)
>> Bienvenido a la Feria de Isla de Maipo
>> Puestos instalados: 24
```

## 🧩 Bloque 2 — Tipos de datos básicos

- Python distingue varios tipos de datos: `int` (números enteros), `float` (números con decimales), `str` (texto, siempre entre comillas), `bool` (solo `True` o `False`).
- Cada variable tiene un tipo, aunque no lo escribas explícitamente — Python lo detecta solo.
- ⚠️ `"10"` (con comillas) es texto, no un número, aunque se vea igual — Python lo trata distinto.

**Ejemplo resuelto:**

```python
duracion_cancion_minutos = 3.5
numero_de_pista = 7
nombre_artista = "Mon Laferte"
es_lanzamiento_nuevo = True
print("Pista", numero_de_pista, "-", nombre_artista)
print("Duración:", duracion_cancion_minutos, "minutos")
print("¿Es lanzamiento nuevo?", es_lanzamiento_nuevo)
>> Pista 7 - Mon Laferte
>> Duración: 3.5 minutos
>> ¿Es lanzamiento nuevo? True
```

## 🧩 Bloque 3 — `input()`

- `input()` detiene el programa y espera que la persona escriba algo por teclado.
- Todo lo que devuelve `input()` es texto (`str`), aunque la persona escriba solo números.
- El mensaje entre paréntesis de `input()` es lo que se le muestra a la persona antes de escribir.

**Ejemplo resuelto:**

```python
equipo_favorito = input("¿Cuál es tu equipo favorito? ")
print("Tu equipo favorito es", equipo_favorito)
>> (la persona escribe: Colo-Colo)
```

## 🧩 Bloque 4 — Conversión de tipos con `int()` y `float()`

- Como `input()` siempre devuelve texto, si necesitas operar matemáticamente con ese valor hay que convertirlo primero.
- `int()` convierte a número entero; `float()` convierte a número con decimales.
- ⚠️ Error típico: sumar un texto con un número sin convertir primero — el programa se detiene con un error.

**Ejemplo resuelto:**

```python
edad_texto = input("¿Cuántos años tienes? ")
edad = int(edad_texto)
edad_en_5_anos = edad + 5
print("En 5 años tendrás", edad_en_5_anos, "años")
>> (la persona escribe: 17)
>> En 5 años tendrás 22 años
```

## 🧩 Bloque 5 — Combinando todo: `input()` + conversión + operaciones

- Ya puedes: pedir datos con `input()`, convertirlos al tipo correcto, guardarlos en variables bien nombradas, y operar con ellos usando `print()` para mostrar resultados.
- Un programa completo normalmente combina varios de estos pasos en orden: primero pedir datos, después operar, al final mostrar.

**Ejemplo resuelto:**

```python
entradas_vendidas_por_hora = int(input("¿Cuántas entradas vendes por hora en la kermés? "))
horas_de_venta = 4
entradas_totales = entradas_vendidas_por_hora * horas_de_venta
print("Total estimado de entradas vendidas:", entradas_totales)
>> (la persona escribe: 15)
>> Total estimado de entradas vendidas: 60
```
