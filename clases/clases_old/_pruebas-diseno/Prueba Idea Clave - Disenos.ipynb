## 💡 La función `input()` y su tipo de dato

Cuando usamos `input()` para pedirle un dato a la persona que usa el programa, Python siempre guarda lo que escribió como texto (`str`) — sin importar si escribió letras o números.

> **Idea clave:** `input()` **siempre** devuelve texto (`str`), nunca un número — aunque la persona escriba puros dígitos.

```python
edad_texto = input("¿Cuántos años tienes? ")
print("Lo que guardé es:", edad_texto)
print("¿Es de tipo str?", type(edad_texto) == str)
# >> ¿Cuántos años tienes? 17
# >> Lo que guardé es: 17
# >> ¿Es de tipo str? True
```

---

## 💡 El ciclo `while`

El ciclo `while` repite un bloque de instrucciones mientras una condición se mantenga verdadera.

> **Idea clave:** la condición del `while` se revisa **antes** de cada vuelta — si parte en falso, el bloque no se ejecuta ni una vez.

```python
intentos = 0
while intentos < 3:
    print("Intento número", intentos + 1)
    intentos = intentos + 1
# >> Intento número 1
# >> Intento número 2
# >> Intento número 3
```

---

## 💡 La condición del `if`

El `if` evalúa una condición y solo ejecuta el bloque indentado que sigue cuando esa condición es verdadera; si es falsa, ese bloque se salta completo.

> **Idea clave:** si la condición del `if` es falsa, Python **se salta todo el bloque** — ni una línea de adentro se ejecuta.

```python
saldo = 5000
if saldo >= 8000:
    print("Alcanza para la entrada")
# (no imprime nada: la condición fue falsa)
```
