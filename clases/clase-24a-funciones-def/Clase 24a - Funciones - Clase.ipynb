## 2️⃣ Introducción al Contenido Nuevo

Estudia cada concepto, ejecuta los ejemplos y observa el resultado:


### 💡 Qué es una función y por qué usarla

un bloque de código con nombre propio que agrupa una tarea, para no repetirla escrita completa cada vez que se necesita.

**Idea clave:** _una función se define una vez y se usa todas las veces que haga falta._


```python
# Sin función: el mismo cálculo repetido
print(8000 - 1500)
print(5000 - 1500)
print(12000 - 1500)
```

### 💡 Definir con `def` y parámetros

`def nombre_funcion(parametro1, parametro2):` abre el bloque; el cuerpo va indentado debajo. Los parámetros son variables que la función espera recibir.

**Idea clave:** _definir la función no la ejecuta — solo queda "guardada la receta" hasta que alguien la llama._


```python
def descuento_socio(precio, es_socia):
    if es_socia:
        return precio - 1500
    return precio
```

### 💡 Llamar la función: argumentos

llamar la función es escribir su nombre con paréntesis y los valores reales (argumentos) que ocupan el lugar de los parámetros.

**Idea clave:** _parámetro es el nombre en la definición; argumento es el valor que se manda al llamar._


```python
precio_final = descuento_socio(8000, True)
print("Paga:", precio_final)
```

### 💡 `return`: devolver un resultado

`return` entrega un valor de vuelta a quien llamó la función y termina la ejecución en ese punto. Sin `return`, la función devuelve `None`.

**Idea clave:** _`print()` muestra algo en pantalla, pero solo `return` deja el valor disponible para usarlo después._


```python
def saluda(nombre):
    print("Hola,", nombre)

resultado = saluda("Camila")
print(resultado)
```

#### ⚠️ Errores típicos a evitar

| Error | Qué ocurre | Cómo corregirlo |
|---|---|---|
| Llamar la función sin paréntesis (`descuento_socio`) | No ejecuta nada, solo referencia a la función | Agregar `()` con los argumentos |
| Confundir `print()` interno con `return` | La función "muestra" el resultado pero no lo entrega — la variable que la recibe queda en `None` | Usar `return` si el valor se necesita después |
| Esperar que una variable definida dentro de la función exista afuera | `NameError` al intentar usarla fuera | Devolver el valor con `return` y guardarlo en una variable afuera |
| Escribir código después de un `return` esperando que se ejecute | Ese código nunca corre | Recordar que `return` corta la función ahí mismo |
