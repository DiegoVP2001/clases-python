**Ítem 1A.1** (4 pts)

Para la feria científica del liceo, un grupo programó la cuenta regresiva del lanzamiento de su cohete a escala: debe imprimir los números del 10 al 1, uno por línea, terminando con "¡Despegue!".

```python
for numero in range(10, 0, -1):
    print(numero)

print("¡Despegue!")

# Esperado:
# 10
# 9
# 8
# ...
# 1
# ¡Despegue!
```

**Ítem 1A.2** (4 pts)

En el club de atletismo, un corredor entrena dando 10 vueltas a la pista, numeradas del 1 al 10. Cada vuelta múltiplo de 3, el entrenador le toma el tiempo con cronómetro. El programa debe contar cuántas veces le tomó el tiempo en total.

```python
tomas_de_tiempo = 0

for vuelta in range(1, 11):
    if vuelta % 3 == 0:
        tomas_de_tiempo = tomas_de_tiempo + 1

print("Tomas de tiempo:", tomas_de_tiempo)

# Esperado:
# Tomas de tiempo: 3
```

🔎 **Qué se revisó:** Que la condición usara el resto de la división (múltiplos de 3) y que el contador se actualizara solo cuando correspondía.

En el taller de robótica se arma una cuadrícula de prueba de sensores de 3 filas por 3 columnas: el sistema debe imprimir en qué fila y columna está cada sensor.

```python
for fila in range(3):
    for columna in range(3):
        print("Sensor en fila", fila, "columna", columna)

# Esperado:
# Sensor en fila 0 columna 0
# Sensor en fila 0 columna 1
# Sensor en fila 0 columna 2
# Sensor en fila 1 columna 0
# Sensor en fila 1 columna 1
# Sensor en fila 1 columna 2
# Sensor en fila 2 columna 0
# Sensor en fila 2 columna 1
# Sensor en fila 2 columna 2
```

🔎 **Qué se revisó:** Que el ciclo interno recorriera las 3 columnas completas, no solo 2.

**Ítem 1A.4** (4 pts)

El estacionamiento del liceo tiene 5 cupos, numerados del 1 al 5. El sistema debe registrar la entrada de vehículos exactamente hasta llenar el estacionamiento, sin registrar ninguno de más.

```python
cupos_ocupados = 0
cupos_totales = 5

while cupos_ocupados < cupos_totales:
    cupos_ocupados = cupos_ocupados + 1
    print("Vehículo N°", cupos_ocupados, "estacionado.")

# Esperado:
# Vehículo N° 1 estacionado.
# Vehículo N° 2 estacionado.
# Vehículo N° 3 estacionado.
# Vehículo N° 4 estacionado.
# Vehículo N° 5 estacionado.
```

🔎 **Qué se revisó:** Que la condición del `while` cortara exactamente al llenar los 5 cupos, sin registrar uno de más.

**Ítem 1B.4** (4 pts)

### Ejercicio 1 — Ventas del almacén (15 pts)

El almacén de barrio de Isla de Maipo va a registrar las ventas de sus primeras 5 transacciones del día. El programa debe pedir cada venta y, al terminar, informar el total vendido y cuántas ventas individuales superaron los \$5.000.

**El programa debe:**
- Usar un `for` con `range(5)` para repetir exactamente 5 veces.
- Por cada vuelta del `for`, pedir la venta correspondiente con `input()` (número entero).
- Acumular el **total vendido** en el día.
- Contar cuántas ventas fueron **superiores a \$5.000**.
- Imprimir ambos resultados con etiqueta al terminar.

<table>
<tr>
  <th>Ejemplo 1</th>
  <th>Ejemplo 2</th>
</tr>
<tr>
  <td>📥 <em>El usuario ingresa</em><pre>3000<br>5500<br>12000<br>4200<br>8000</pre></td>
  <td>📥 <em>El usuario ingresa</em><pre>1000<br>2000<br>3000<br>1500<br>900</pre></td>

<tr>
  <td>📤 <em>El programa imprime</em><pre>Total vendido: 32700
Ventas sobre $5.000: 3</pre></td>
  <td>📤 <em>El programa imprime</em><pre>Total vendido: 8400

```python
total_vendido = 0
ventas_sobre_umbral = 0

for numero_venta in range(5):
    venta = int(input("Ingresa el monto de la venta: "))
    total_vendido = total_vendido + venta
    if venta > 5000:
        ventas_sobre_umbral = ventas_sobre_umbral + 1

print("Total vendido:", total_vendido)
print("Ventas sobre $5.000:", ventas_sobre_umbral)
```

🔎 **Qué se revisó:** Que se pidiera exactamente 5 ventas con `input()`, y que el total y el conteo sobre el umbral se acumularan correctamente vuelta a vuelta.

### Ejercicio 2 — Formación del baile de Aniversario (17 pts)

Para el número de baile del Aniversario, cada integrante de una fila debe gritar el número de su fila: la fila 1 tiene 1 integrante, la fila 2 tiene 2, y así sucesivamente hasta la fila indicada. Además, la profesora a cargo quiere saber cuántos integrantes participan en total en el número completo.

**El programa debe:**
- Usar `filas = 4`.
- Para cada fila (empezando en 1), imprimir el número de esa fila una vez por cada integrante que le corresponde (una impresión por línea).
- Contar el **total de integrantes** que participan en el número completo (la suma de todas las filas) e imprimirlo al final, con etiqueta.

📤 <em>El programa imprime:</em>
<pre>1
2
2
3
3
3
4
4
4
4
Total de integrantes: 10</pre>

```python
filas = 4
total_integrantes = 0

for fila in range(1, filas + 1):
    for integrante in range(fila):
        print(fila)
        total_integrantes = total_integrantes + 1

print("Total de integrantes:", total_integrantes)
```

🔎 **Qué se revisó:** Que el ciclo anidado repitiera cada fila la cantidad correcta de veces, y que el acumulador de integrantes sumara en cada impresión, no solo al final.

### Ejercicio 3 — Estación meteorológica (18 pts)

Una estación meteorológica escolar recibe temperaturas ingresadas manualmente durante el día. El programa debe seguir pidiendo temperaturas (números que pueden tener decimales) hasta que se ingrese **-999**, que marca el fin del registro, y luego mostrar el promedio del día.

**El programa debe:**
- Pedir temperaturas repetidamente con `while`, hasta recibir `-999`.
- Acumular la suma y contar cuántas temperaturas válidas se ingresaron (sin contar el -999).
- Imprimir el **promedio** al terminar.
- Si no se ingresó ninguna temperatura válida antes del -999, imprimir un mensaje indicando que no hay datos, sin calcular el promedio.

<table>
<tr>
  <th>Ejemplo 1</th>
  <th>Ejemplo 2</th>
</tr>
<tr>
  <td>📥 <em>El usuario ingresa</em><pre>18.5<br>20.0<br>19.2<br>-999</pre></td>
  <td>📥 <em>El usuario ingresa</em><pre>-999</pre></td>
</tr>
<tr>
  <td>📤 <em>El programa imprime</em><pre>Promedio del día: 19.23</pre></td>
  <td>📤 <em>El programa imprime</em><pre>No se registraron temperaturas.</pre></td>
</tr>
</table>

```python
suma_temperaturas = 0
cantidad_temperaturas = 0

temperatura = float(input("Ingresa una temperatura (o -999 para terminar): "))

while temperatura != -999:
    suma_temperaturas = suma_temperaturas + temperatura
    cantidad_temperaturas = cantidad_temperaturas + 1
    temperatura = float(input("Ingresa una temperatura (o -999 para terminar): "))

if cantidad_temperaturas > 0:
    promedio = suma_temperaturas / cantidad_temperaturas
    print("Promedio del día:", promedio)
else:
    print("No se registraron temperaturas.")
```

🔎 **Qué se revisó:** Que el promedio se calculara con la suma y el conteo correctos, y que el caso sin ninguna temperatura ingresada no intentara dividir por cero.

### Ejercicio 4 — Caja fuerte del escape room (18 pts)

En el escape room del liceo, una caja fuerte tiene un código de 4 dígitos. Cada equipo dispone de **como máximo 5 intentos**; el programa debe cortar apenas alguien acierte el código, o informar que se acabaron los intentos si nadie lo logra.

**El programa debe:**
- Usar `codigo_correcto = 4271` y `intentos_maximos = 5`.
- Pedir un intento (número entero) repetidamente, contando cuántos van.
- Si el intento es correcto, felicitar y **cortar el ciclo** de inmediato (sin seguir pidiendo intentos).
- Si se agotan los 5 intentos sin acertar, informar que la caja quedó bloqueada.

<table>
<tr>
  <th>Ejemplo 1</th>
  <th>Ejemplo 2</th>
</tr>
<tr>
  <td>📥 <em>El usuario ingresa</em><pre>1234<br>4321<br>4271</pre></td>
  <td>📥 <em>El usuario ingresa</em><pre>1111<br>2222<br>3333<br>4444<br>5555</pre></td>
</tr>
<tr>
  <td>📤 <em>El programa imprime</em><pre>¡Código correcto! Caja abierta en el intento 3</pre></td>
  <td>📤 <em>El programa imprime</em><pre>Caja bloqueada: se agotaron los 5 intentos.</pre></td>
</tr>
</table>

```python
codigo_correcto = 4271
intentos_maximos = 5
intento_actual = 0
acerto = False

while intento_actual < intentos_maximos:
    intento_actual = intento_actual + 1
    intento = int(input("Ingresa el código: "))
    if intento == codigo_correcto:
        acerto = True
        break

if acerto:
    print("¡Código correcto! Caja abierta en el intento", intento_actual)
else:
    print("Caja bloqueada: se agotaron los 5 intentos.")
```

🔎 **Qué se revisó:** Que el ciclo cortara apenas se acertara el código, y que el mensaje final distinguiera correctamente entre acierto y agotamiento de intentos.
