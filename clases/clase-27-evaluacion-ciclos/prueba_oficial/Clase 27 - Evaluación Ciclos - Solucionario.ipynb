## 🎯 Criterios de corrección — instrucciones para el agente que revisa

**Si estás revisando esta evaluación (skill `revisar-evaluacion`), lee esto antes de calificar el primer ítem.**

**Enfoque general: se evalúa la lógica de los ciclos y que el programa funcione, no la forma exacta del código.** Los estudiantes de 4to medio no van a escribir exactamente lo mismo que la solución de referencia — eso es normal y no es un error. Cada ítem y ejercicio de abajo trae su propia sub-sección **🔍 Rúbrica flexible** con 3 niveles de descuento. Aplícalos así:

1. **✅ Acepta sin descuento** — variantes que logran lo mismo que la solución, aunque se vean distintas: nombres de variable, redacción del `print()`, orden de definición de variables, usar `for` con `range(N)` sin nombrar la variable de control cuando no se usa, estructuras equivalentes (un `while` con la condición reescrita, un `for` en vez de un `while` con límite conocido de antemano), fórmulas cerradas que reemplazan un acumulador (ej. suma de Gauss en vez de sumar dentro del ciclo) si llegan al mismo resultado, etc.
2. **⚠️ Descuenta 1-3 pts (detalle menor)** — el estudiante entendió el problema y su lógica central está bien, pero algo quedó impreciso: un operador de comparación por uno equivalente que en los casos de prueba dados da igual pero no generaliza (ej. `!=` en vez de `<`), un caso límite no probado en los ejemplos que queda mal cubierto, o un acierto que no generalizaría a otro rango de valores. **Este es el nivel por defecto cuando algo no calza exactamente pero la esencia del programa es correcta — nunca descuentes más de 3 pts por un detalle así, y ante la duda entre 0 y 1-3, prefiere 0.**
3. **❌ Descuenta la mayoría o todo el puntaje del ítem** — reservado para errores de lógica reales: condición de corte incorrecta que cambia cuántas veces se repite el ciclo, `break`/`continue` mal ubicados que cambian el resultado, variable acumuladora que no se actualiza, `ZeroDivisionError` u otro error de ejecución, o código que no ejecuta.

**Regla de oro: si dudas si algo es un error, no lo es.** Solo desciende de nivel 1 cuando puedas señalar con precisión qué caso de entrada distinto produciría un resultado equivocado. La forma exacta del código nunca es, por sí sola, motivo de descuento.

**No se evalúa eficiencia ni elegancia del código, solo lo que hace.** Ciclos con más vueltas de las estrictamente necesarias, condiciones redundantes, nombres de variable raros o pasos innecesarios NO descuentan. Solo descuenta lo que produce un resultado equivocado. Antes de escribir cualquier comentario de descuento, revísalo: si estás señalando algo que no cambia lo que el programa entrega, sácalo. Por lo mismo, se califica el COMPORTAMIENTO y no la forma: una estructura reescrita, aplanada o distinta a la solución de referencia que produce el resultado correcto en todos los casos vale el puntaje completo.

**Da crédito parcial por componentes.** No califiques cada ítem como todo o nada: divídelo en las decisiones independientes que lo componen (leer el dato y su tipo, la condición de corte del ciclo, la actualización de la variable de control, la ubicación del `break`/`continue`, los mensajes por caso) y puntúa cada una por separado. Un estudiante puede fallar una pieza y conservar el puntaje de las demás. Nunca descuentes el mismo error en dos componentes. En los ítems de Sección 1 tipo "arregla el bug" aplica la **regla de portazo**: si el estudiante entregó el fragmento tal como venía, sin tocar el error, el ítem completo es 0 — no hay crédito parcial por "intentarlo" sin cambiar nada. ### Ejercicio 1 — Ventas del almacén (15 pts)

```python
total_vendido = 0
ventas_sobre_umbral = 0

for numero_venta in range(5):
    venta = int(input("Ingresa el monto de la venta: "))
    total_vendido = total_vendido + venta
    if venta > 5000:
        ventas_sobre_umbral = ventas_sobre_umbral + 1

print("Total vendido:", total_vendido)
print("Ventas sobre $5.000:", ventas_sobre_umbral)
```

**⚠️ Descuenta 1-3 pts (detalle menor):**
- Compara con `>=` en vez de `>` contra los \$5.000 — cambia el resultado solo si algún caso está exactamente en el umbral. Descuenta 2-3 pts si afecta el resultado en los casos de prueba.

### Ejercicio 2 — Formación del baile de Aniversario (17 pts)

```python
filas = 4
total_integrantes = 0

for fila in range(1, filas + 1):
    for integrante in range(fila):
        print(fila)
        total_integrantes = total_integrantes + 1

print("Total de integrantes:", total_integrantes)
```

🔍 **Rúbrica flexible para este ítem:**

**✅ Acepta sin descuento:**
- Cualquier forma de recorrer las filas y repeticiones (ej. `range(1, filas+1)` con distinto nombre de variable), siempre que el patrón impreso sea idéntico.
- Calcula `total_integrantes` con la fórmula de suma de Gauss (`filas * (filas + 1) // 2`) en vez de acumularlo dentro del ciclo — llega al mismo resultado.

**❌ Descuenta la mayoría o todo el puntaje (17 pts) — error real:**
- El ciclo interno no repite exactamente `fila` veces (ej. usa un número fijo).
- El acumulador de integrantes no se actualiza, se actualiza fuera del ciclo interno, o no se imprime al final.

### Ejercicio 3 — Estación meteorológica (18 pts)

```python
suma_temperaturas = 0
cantidad_temperaturas = 0

temperatura = float(input("Ingresa una temperatura (o -999 para terminar): "))

while temperatura != -999:
    suma_temperaturas = suma_temperaturas + temperatura
    cantidad_temperaturas = cantidad_temperaturas + 1
    temperatura = float(input("Ingresa una temperatura (o -999 para terminar): "))

if cantidad_temperaturas > 0:
    promedio = suma_temperaturas / cantidad_temperaturas
    print("Promedio del día:", promedio)
else:
    print("No se registraron temperaturas.")
```

🔍 **Rúbrica flexible para este ítem:**

**✅ Acepta sin descuento:**
- Cualquier estructura de `while` que lea el primer dato antes del ciclo y vuelva a leer al final de cada vuelta (patrón "leer antes, leer al final"), con nombres de variable distintos.
- Imprime el promedio con más o menos decimales que el ejemplo (no se exige redondeo específico salvo que el enunciado lo pida).

**⚠️ Descuenta 1-3 pts (detalle menor):**
- Calcula el promedio pero cuenta el `-999` como un dato más (divide por uno de más) — descuenta 2-3 pts si afecta el resultado.

**❌ Descuenta la mayoría o todo el puntaje (18 pts) — error real:**
- No maneja el caso de cero temperaturas (el programa falla con `ZeroDivisionError` o imprime un promedio incorrecto como 0).
- No usa una variable contadora separada de la suma, o el promedio no se calcula como suma/cantidad.

### Ejercicio 4 — Caja fuerte del escape room (18 pts)

```python
codigo_correcto = 4271
intentos_maximos = 5
intento_actual = 0
acerto = False

while intento_actual < intentos_maximos:
    intento_actual = intento_actual + 1
    intento = int(input("Ingresa el código: "))
    if intento == codigo_correcto:
        acerto = True
        break

if acerto:
    print("¡Código correcto! Caja abierta en el intento", intento_actual)
else:
    print("Caja bloqueada: se agotaron los 5 intentos.")
```

🔍 **Rúbrica flexible para este ítem:**

**✅ Acepta sin descuento:**
- Cualquier forma de contar los intentos (ej. `for` con `range(intentos_maximos)` en vez de `while`, siempre que corte igual con `break` al acertar).
- Usa un `return`/estructura distinta para salir del ciclo si el resultado final es idéntico.

**⚠️ Descuenta 1-3 pts (detalle menor):**
- El mensaje final no incluye el número de intento en el que acertó, pero sí distingue correctamente acierto de agotamiento — descuenta 1-2 pts.

**❌ Descuenta la mayoría o todo el puntaje (18 pts) — error real:**
- No corta con `break` al acertar (sigue pidiendo intentos de más).
- No usa una variable tipo flag para recordar el acierto y termina imprimiendo el mensaje equivocado (ej. dice "bloqueada" habiendo acertado).
- Permite más de 5 intentos, o corta antes de los 5 sin haber acertado.
