# N°35 — Listas: creación y acceso

**Objetivo de la sesión:** Convertir una fila de texto de `guaguas` en lista con `split(",")` y acceder a sus datos por índice y slicing.

## ✅ Contenido que debe quedar consolidado al cierre

- `variable.split(",")` convierte un texto en lista
- Acceder a un elemento por índice (`lista[0]`, `lista[-1]`)
- Slicing (`lista[1:3]`)
- Puente explícito con Clase 32 (`split()` en strings)

### 💻 Lo que se espera que un estudiante escriba (ejemplo mínimo, con `guaguas`)

```python
fila_texto = "1920,María,F,1234"
datos_guagua = fila_texto.split(",")

print("Año:", datos_guagua[0])
print("Nombre:", datos_guagua[1])
print("Los primeros dos datos:", datos_guagua[:2])
```
