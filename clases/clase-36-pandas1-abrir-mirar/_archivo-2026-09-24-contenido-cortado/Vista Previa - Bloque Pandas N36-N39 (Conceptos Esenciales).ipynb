## N°36 — Pandas 1: abrir y mirar

**Objetivo:** abrir un CSV real, mirarlo, elegir columnas, y contar/filtrar vacíos.

```python
import pandas as pd

tabla = pd.read_csv("guaguas_maquillada.csv")
print("Filas y columnas:", tabla.shape)
print("Columnas:", list(tabla.columns))
tabla.head()
```

```python
columnas_elegidas = tabla[["nombre", "sexo"]]
columnas_elegidas.head()
```

```python
print("Nulos por columna:")
print(tabla[["nombre", "sexo"]].isna().sum())

tabla_sin_nulos_sexo = tabla[tabla["sexo"].notna()]
print("Filas después de filtrar sexo nulo:", len(tabla_sin_nulos_sexo))
```

## Bloque "limpiar, normalizar y filtrar" — repartido en 2 clases (confirmado por Diego)

**Clase 1 (A+C):** las dos técnicas "directas" de limpiar a número — limpiarlo y descartar lo que no se pudo limpiar.
**Clase 2 (B+D):** las dos técnicas de categorías — normalizar (el regalo) y filtrar por valor.

⚠️ **Ajuste de código respecto a la versión anterior:** "descartar filas sucias" antes revisaba `cantidad_guaguas` Y `sexo_normalizado` a la vez, pero `sexo_normalizado` no existe todavía en la Clase 1 (se crea recién en la Clase 2). Se separó en dos pasos: cada clase descarta solo lo que ensució *su propia* columna, y el descarte combinado (las dos columnas juntas) se hizo el primer paso de la clase de derivar/agrupar/ordenar, como puente entre los dos bloques.

### Clase 1 (A+C) — Limpiar a número + descartar filas sucias

```python
tabla["cantidad_guaguas"] = pd.to_numeric(
    tabla["n"].astype(str).str.strip().str.replace(".", "", regex=False), errors="coerce"
)
tabla[["n", "cantidad_guaguas"]].head()
```

```python
tabla_sin_nulos_numero = tabla[tabla["cantidad_guaguas"].notna()]
print("Filas antes:", len(tabla))
print("Filas después de descartar cantidad_guaguas nula:", len(tabla_sin_nulos_numero))
```

### Clase 2 (B+D) — Normalizar categorías (el regalo) + filtrar por valor

```python
# El regalo: contar "F" a secas no capta las filas "Femenino"
print("Valores de sexo tal como vienen:", tabla["sexo"].unique())

cantidad_ninas_sin_normalizar = (tabla["sexo"] == "F").sum()
print("Niñas contadas sin normalizar:", cantidad_ninas_sin_normalizar)
```

```python
equivalencias_sexo = {"femenino": "F", "masculino": "M", "f": "F", "m": "M"}
tabla["sexo_normalizado"] = tabla["sexo"].astype(str).str.strip().str.lower().replace(equivalencias_sexo)

cantidad_ninas_normalizado = (tabla["sexo_normalizado"] == "F").sum()
print("Niñas contadas después de normalizar:", cantidad_ninas_normalizado)
```

```python
tabla_nombre_filtrado = tabla[tabla["nombre"] == "Antonia"]
tabla_nombre_filtrado.head()
```

## N°38 (tentativo) — derivar, agrupar y ordenar

**Objetivo:** juntar las dos limpiezas de las clases anteriores, crear una columna derivada, agrupar y ordenar/cortar — la operación que responde la pregunta del proyecto.

```python
# Puente: recién acá se combinan las dos limpiezas (número + categoría) de las dos clases anteriores
tabla_limpia = tabla[tabla["cantidad_guaguas"].notna() & tabla["sexo_normalizado"].notna()].copy()
tabla_limpia.head()
```

```python
tabla_limpia["decada"] = (tabla_limpia["anio"] // 10) * 10
tabla_limpia[["anio", "decada"]].head()
```

```python
resumen = tabla_limpia.groupby(["decada", "nombre"])["cantidad_guaguas"].sum().reset_index()
resumen.head()
```

```python
nombre_lider_por_decada = (
    resumen.sort_values("cantidad_guaguas", ascending=False).groupby("decada").head(1)
)
nombre_lider_por_decada
```

## N°39 — Pandas 3: graficar

**Objetivo:** `.plot(kind="bar")` para comparar categorías y `.plot(kind="line")` para ver tendencia — ambos a fondo, con título y ejes.

```python
import matplotlib.pyplot as plt

total_por_decada = tabla_limpia.groupby("decada")["cantidad_guaguas"].sum()
total_por_decada.plot(kind="bar", title="Total de guaguas nacidas por década")
plt.xlabel("Década"); plt.ylabel("Cantidad de guaguas"); plt.show()
```

```python
nombre_lider_por_decada.plot(
    x="decada", y="cantidad_guaguas", kind="line", marker="o",
    title="Cantidad de guaguas del nombre líder, por década",
)
plt.xlabel("Década"); plt.ylabel("Cantidad con el nombre líder"); plt.show()
```
