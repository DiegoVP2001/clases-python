**Objetivo de la sesión:** Recorrer listas paralelas (columnas de `guaguas`) con `for` para contar, sumar/promediar, encontrar el máximo, y filtrar construyendo una lista nueva con `append`.

## ✅ Contenido que debe quedar consolidado al cierre

- Filtrar a mano: `for` + `if` + `append()` a una lista nueva
- **De regalo:** al contar "cuántas niñas" con `if valor == "F"`, notan que el conteo da mal porque hay filas `"Femenino"` — siembra por qué hace falta normalizar categorías

### 💻 Lo que se espera que un estudiante escriba (ejemplo mínimo, con `guaguas`)

```python
sexos = ["F", "F", "Femenino", "M", "F"]
cantidad_ninas = 0
for sexo in sexos:
    if sexo == "F":
        cantidad_ninas += 1

print("Niñas contadas:", cantidad_ninas, "de", len(sexos), "filas")
# ¿Por qué no da 4, si hay 4 niñas en la lista?
```
